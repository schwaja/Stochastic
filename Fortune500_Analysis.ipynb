```python
# Q1: Simple Linear Regression - Revenues vs Profits
print("\n" + "="*70)
print("LINEAR REGRESSION ANALYSIS")
print("="*70)

# Prepare data
data_clean = df[['Revenues', 'Profits', 'Assets', 'Market Value', 'Number of Employees']].dropna()
X_q1 = data_clean[['Revenues']].values
y_q1 = data_clean['Profits'].values

# Fit model
model_q1 = LinearRegression()
model_q1.fit(X_q1, y_q1)

# Predictions
y_pred_q1 = model_q1.predict(X_q1)
r2_q1 = r2_score(y_q1, y_pred_q1)
rmse_q1 = np.sqrt(mean_squared_error(y_q1, y_pred_q1))
mae_q1 = mean_absolute_error(y_q1, y_pred_q1)

print("\n*** Q1: PROFITABILITY VS. SIZE RELATIONSHIP ***")
print(f"Model: Profits = {model_q1.intercept_:.2f} + {model_q1.coef_[0]:.4f} × Revenues")
print(f"R² Score: {r2_q1:.4f} (explains {r2_q1*100:.2f}% of variance)")
print(f"RMSE: ${rmse_q1:,.0f}")
print(f"MAE: ${mae_q1:,.0f}")

# Statistical significance test
correlation_r = df['Revenues'].corr(df['Profits'])
n = len(data_clean)
t_stat = correlation_r * np.sqrt(n - 2) / np.sqrt(1 - correlation_r**2)
p_value = stats.t.sf(abs(t_stat), n - 2) * 2
print(f"Correlation (Pearson r): {correlation_r:.4f}")
print(f"P-value: {p_value:.2e} (Highly Significant)" if p_value < 0.001 else f"P-value: {p_value:.4f}")
```
