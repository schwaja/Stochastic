```python
# Data cleaning: Convert text columns to numeric
def clean_numeric(value):
    """Convert formatted strings to numeric values"""
    if isinstance(value, str):
        # Remove currency signs, commas, percentage signs
        value = value.replace('$', '').replace(',', '').replace('%', '')
        try:
            return float(value)
        except:
            return np.nan
    return value

# Identify numeric columns to clean
numeric_cols = ['Number of Employees', 'Revenues', 'Revenue Change', 'Profits', 
                'Profit Change', 'Assets', 'Market Value']

# Clean numeric columns
for col in numeric_cols:
    if col in df.columns:
        df[col] = df[col].apply(clean_numeric)

print("Data cleaned successfully!")
print("\nMissing Values:")
print(df.isnull().sum())
print("\nSummary Statistics:")
print(df.describe())
```
